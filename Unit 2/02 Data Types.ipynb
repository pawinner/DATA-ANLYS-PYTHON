## Numbers
- `int` = intergers - จำนวนเต็ม
- `float` = real numbers - ทศนิยม ok
- `complex` = จำนวนเชิงซ้อน

```python
print(67)
print(676967.69)
print(1.00000)
```

Use `type()` to check

```python
print(type(67))
print(type(676967.69))
print(type(1.00000))
```

Complex numbers: a+bj

```python
print(6+7j)
print(type(6+7j))
```


## Strings
In Python, Strings = `str`. Can use single, double, triple quotes.

```python
print('Skibidi')
print("Skibidii")
print('''Skibidiii''')
print(type('''Skibidiii'''))
```

What if we actually want the quote symbols as the text? - Use different quotes!

```python
print('She said "ว้อดส์"')
print('''It's ok to say "hit the road dude".''''')
```

or use escape characters (too much bruh;)

```python
print("She said \"ว้อดส์\"")
print("It's ok to say \"hit the road dude\".")
print("Yoooo\\")
```

```python
print("1st line.\n2nd line.\n3rd line")
```

```python
print("Pseu" "doephe" "drine")
print("Pseudo" "ephedrine")
print("P" "seudoephedr" "ine")
```
