# Operators

## Arithmetic Operators
#### ` + , - , * , / `

```python
a = 20
b = 5
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```python
a = 2
print(-a)
b = -5
print(-b)
```

#### Modulus
`a % b` = a mod b

```python
x = 25
print(x % 4) # Since 25 = 4*6 + 1
print(x % 5)
print(x % 7)
```

```python
print(2 % 0)
```

### Exponential: ยกกำลัง

`a ** b`

```python
print(2*5)
print(2**5)
print(2**0)
print(2**-1)
```

### Floor Division: หารแล้วปัดลง
`//`

```python
print(32/3)
print(32//3)
print(-32/3)
print(-32//3)
```

## Assignment Operators

### Simple

```python
a = 3
print(a)
b = 4
a = b + 1
print(a)
```

### Compound

```python
a = 2
print(a)
a = a + 5
print(a)
```

```python
a = 2
print(a)
a += 5
print(a)
```

```python
a=b=c=d=e=f=67
a -= 5
print(a)
a *= 5
print(a)
a /= 5
print(a)
a %= 5
print(a)
a **= 5
print(a)
a //= 5
print(a)
```

## Comparison Operators

`True` or `False`

```python
a = 20
print(a == 20)
print(a == 15)
print(a != 20)
print(a != 15)
print(a > 20)
print(a < 20)
print(a >= 20)
print(a <= 20)
```

## Logical Operators


`and, or, not`

```python
p = True
q = False
print(p and q)
print(p or q)
print(not q)
```

```python
username = input("Your name: ")
print(username == "Win" or username == "Sigma")
```
